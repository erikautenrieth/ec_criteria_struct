```python
import os
import re

def process_file(file_path, output_folder):
    with open(file_path, 'r', encoding="utf-8") as file:
        lines = file.readlines()

    processed_lines = []
    for i, line in enumerate(lines):
        line = re.sub(r'\[NOT\]', ' [AND1] [NOT]', line)

        if i < len(lines) - 1:  # Not the last line
            processed_lines.append(line.strip() + ' [AND1]\n')
        else:  # Last line
            processed_lines.append(line.strip() + '\n')

    os.makedirs(output_folder, exist_ok=True)

    output_file = os.path.join(output_folder, os.path.basename(file_path))
    with open(output_file, 'w', encoding="utf-8") as file:
        file.writelines(processed_lines)


input_folder = 'lct_p1_half'  
output_folder = 'lct_p1_half_with_and'  

for filename in os.listdir(input_folder):
    file_path = os.path.join(input_folder, filename)
    process_file(file_path, output_folder)
```

### Find files with most and least operators

```python
import os
import re
from collections import Counter

def count_operators(text):
    return {
        'AND': len(re.findall(r'\[AND\]', text)),
        'OR': len(re.findall(r'\[OR\]', text)),
        'NOT': len(re.findall(r'\[NOT\]', text))
    }

def analyze_files(directory):
    operator_counts = Counter()
    files_with_all_operators = []

    for filename in os.listdir(directory):
        if filename.endswith('.txt') and filename.startswith('NCT'):
            with open(os.path.join(directory, filename), 'r', encoding="utf-8") as file:
                content = file.read()
                counts = count_operators(content)
                total_count = sum(counts.values())
                if total_count > 0:
                    operator_counts[filename] = total_count
                    if all(counts.values()):  # Check if all operators are present
                        files_with_all_operators.append((filename, total_count))

    if not operator_counts:
        print("Keine gültigen Dateien gefunden.")
        return

    print("Top 5 Dateien mit den meisten Operatoren:")
    for file, count in operator_counts.most_common(5):
        print(f"{file.split('.')[0]}: {count} Operatoren")

    print("\n5 Dateien mit den wenigsten Operatoren (mindestens je ein AND, OR und NOT):")
    files_with_all_operators.sort(key=lambda x: x[1])
    for file, count in files_with_all_operators[:5]:
        print(f"{file.split('.')[0]}: {count} Operatoren")

directory = "dataset/train/output"

analyze_files(directory)
```

```python
import os
import re

def count_operators_for_files(file_list, directory):
    def count_operators(text):
        return sum([
            len(re.findall(r'\[AND\]', text)),
            len(re.findall(r'\[OR\]', text)),
            len(re.findall(r'\[NOT\]', text))
        ])

    results = []
    for i, filename in enumerate(file_list, 1):
        file_path = os.path.join(directory, filename)
        if os.path.exists(file_path):
            with open(file_path, 'r', encoding="utf-8") as file:
                content = file.read()
                operator_count = count_operators(content)
                results.append((i, filename, operator_count))
        else:
            print(f"Warnung: Datei {filename} nicht gefunden.")

    return results

# Beispielaufruf
shot_list = [
    "NCT03865433.txt",
    "NCT03860324.txt",
    "NCT03860233.txt",
    "NCT03923231.txt",
    "NCT03930121.txt"
]

directory = "dataset/train/output"

results = count_operators_for_files(shot_list, directory)

for num, filename, count in results:
    print(f"{num}. {filename.split('.')[0]}: {count} Operatoren")
```
