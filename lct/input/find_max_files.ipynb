```python
import json
import os
from collections import Counter
## max Files p4

def analyze_clinical_trial_criteria(directory):
    def count_annotations(data):
        counter = Counter()
        
        def traverse(node):
            if isinstance(node, dict):
                for key, value in node.items():
                    if key in ['AND', 'OR', 'NOT']:
                        counter[key] += 1
                    elif isinstance(value, list):
                        counter['entities'] += len(value)
                    traverse(value)
            elif isinstance(node, list):
                for item in node:
                    traverse(item)
        
        traverse(data)
        return counter

    results = []
    
    for filename in os.listdir(directory):
        if filename.endswith(('_inc.txt', '_exc.txt')):
            filepath = os.path.join(directory, filename)
            with open(filepath, 'r', encoding="utf-8") as file:
                try:
                    data = json.load(file)
                    counts = count_annotations(data)
                    total_annotations = sum(counts.values())
                    results.append((filename, total_annotations, counts))
                except json.JSONDecodeError:
                    print(f"Error decoding JSON in file: {filename}")

    results.sort(key=lambda x: x[1], reverse=True)
    
    # Ausgabe der Top 15 Dateinamen
    print("Top 15 Dateien mit den meisten Annotationen:")
    for i, (filename, _, _) in enumerate(results[:15], 1):
        print(f"{i}. {filename}")
    print()

    # Detaillierte Ausgabe für alle Dateien
    print("Detaillierte Informationen für alle Dateien:")
    for filename, total, counts in results:
        print(f"Datei: {filename}")
        print(f"Gesamtanzahl der Annotationen: {total}")
        print(f"AND: {counts['AND']}")
        print(f"OR: {counts['OR']}")
        print(f"NOT: {counts['NOT']}")
        print(f"Entitäten: {counts['entities']}")
        print()
# Beispielaufruf
```

```python
analyze_clinical_trial_criteria('dataset_p4_prompt1_new/train/output')
```
