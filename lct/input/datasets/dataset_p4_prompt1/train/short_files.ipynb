```python
import os

def count_lines_in_file(file_path):
    with open(file_path, 'r', encoding="utf-8") as file:
        return sum(1 for _ in file)

def find_files_with_line_count(folder_path, min_lines=60, max_lines=110):
    matching_files = []

    for filename in os.listdir(folder_path):
        if filename.endswith('.txt'):
            file_path = os.path.join(folder_path, filename)
            line_count = count_lines_in_file(file_path)
            if min_lines <= line_count <= max_lines:
                matching_files.append(filename)

    return matching_files

# Beispielaufruf:
output_folder = 'output'
matching_files = find_files_with_line_count(output_folder)
print("Dateien mit Zeilenanzahl zwischen 60 und 110:", matching_files)
```
