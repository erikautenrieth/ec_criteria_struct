```python
import os

def count_characters_in_folder(folder_path):
    total_characters = 0
    file_count = 0

    # Durchlaufe alle Dateien im angegebenen Ordner
    for filename in os.listdir(folder_path):
        if filename.endswith(".txt"):
            file_path = os.path.join(folder_path, filename)
            try:
                with open(file_path, 'r', encoding='utf-8') as file:
                    content = file.read()
                    char_count = len(content)
                    total_characters += char_count
                    file_count += 1
                    print(f"{filename}: {char_count} Zeichen")
            except Exception as e:
                print(f"Fehler beim Lesen von {filename}: {str(e)}")

    if file_count > 0:
        average_characters = total_characters / file_count
        print(f"\nGesamtanzahl der Dateien: {file_count}")
        print(f"Gesamtanzahl der Zeichen: {total_characters}")
        print(f"Durchschnittliche Anzahl der Zeichen pro Datei: {average_characters:.2f}")
    else:
        print("Keine .txt-Dateien gefunden.")

# Beispielaufruf
folder_path = "example_kriteria"  # Ersetze dies durch den tatsächlichen Pfad zu deinem Ordner
count_characters_in_folder(folder_path)
```
