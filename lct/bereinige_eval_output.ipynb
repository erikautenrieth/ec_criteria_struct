```python
import os
```

```python
def process_files(base_directory):
    eval_directories = [d for d in os.listdir(base_directory) if d.startswith("eval_")]
    print(eval_directories)
    for eval_directory in eval_directories:
        eval_path = os.path.join(base_directory, eval_directory)
        model_output_directories = [d for d in os.listdir(eval_path) if d.startswith("model_output")]
        print(model_output_directories)
        for model_output_directory in model_output_directories:
            model_output_path = os.path.join(eval_path, model_output_directory)
            for root, dirs, files in os.walk(model_output_path):
                if "output" in dirs:
                    output_path = os.path.join(root, "output")
                    print(output_path)
                    print(f"Processing directory: {output_path}")
                    for file in os.listdir(output_path):
                        if file.endswith(".txt"):
                            file_path = os.path.join(output_path, file)
                            process_file(file_path)

def process_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()

    start_tag = "<EDITED_CRITERIA>"
    end_tag = "</EDITED_CRITERIA>"
    start_index = content.find(start_tag)
    end_index = content.find(end_tag)

    if start_index != -1 and end_index != -1:
        start_index += len(start_tag)
        extracted_content = content[start_index:end_index].strip()
        with open(file_path, 'w', encoding='utf-8') as file:
            file.write(extracted_content)
```
