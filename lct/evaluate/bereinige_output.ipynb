```python
import os
```

```python

def process_files(base_directory):
    for root, dirs, files in os.walk(base_directory):
        # Prüfen, ob das Verzeichnis mit "eval_" beginnt
        if os.path.basename(root).startswith("eval_"):
            for subdir in dirs:
                if subdir == "output":
                    output_dir = os.path.join(root, subdir)
                    for file in os.listdir(output_dir):
                        if file.endswith(".txt"):
                            file_path = os.path.join(output_dir, file)
                            process_file(file_path)

def process_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()

    start_tag = "<EDITED_CRITERIA>"
    end_tag = "</EDITED_CRITERIA>"
    start_index = content.find(start_tag)
    end_index = content.find(end_tag)

    if start_index != -1 and end_index != -1:
        # Extrahiere den Inhalt zwischen den Tags
        start_index += len(start_tag)
        extracted_content = content[start_index:end_index].strip()

        # Schreibe den extrahierten Inhalt zurück in die Datei
        with open(file_path, 'w', encoding='utf-8') as file:
            file.write(extracted_content)

```

```python
# Beispielaufruf
base_directory = "/path/to/base_directory"  # Ersetze durch das Verzeichnis, das du durchsuchen möchtest
process_files(base_directory)
```
