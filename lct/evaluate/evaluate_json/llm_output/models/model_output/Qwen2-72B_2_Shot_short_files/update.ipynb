```python
import os

# Verzeichnis, in dem die Dateien umbenannt werden sollen
folder_path = 'output'

# Durchlaufe alle Dateien im Verzeichnis
for filename in os.listdir(folder_path):
    # Prüfe, ob der Dateiname den exakten String "_2_shot" enthält
    if "_2_shot" in filename:
        # Erstelle den neuen Dateinamen, indem "_2_shot" entfernt wird
        new_filename = filename.replace("_2_shot", "")

        # Vollständige Pfade für alte und neue Datei erstellen
        old_file = os.path.join(folder_path, filename)
        new_file = os.path.join(folder_path, new_filename)

        # Datei umbenennen
        os.rename(old_file, new_file)

print("Alle relevanten Dateien wurden erfolgreich umbenannt.")
```
