```python
class Rule:
    def __init__(self, rule_type=None, left=None, right=None, raw_text=None, condition=None, observation=None):
        self.rule_type = rule_type
        self.left = left
        self.right = right
        self.raw_text = raw_text
        self.condition = condition
        self.observation = observation

    def to_dict(self):
        result = {}
        if self.rule_type:
            result[self.rule_type] = {}
            if self.left:
                result[self.rule_type]["left"] = self.left.to_dict()
            if self.right:
                result[self.rule_type]["right"] = self.right.to_dict()
        if self.raw_text:
            result["raw_text"] = self.raw_text
        if self.condition:
            result["Condition"] = self.condition
        if self.observation:
            result["Observation"] = self.observation
        return result

# Beispiel der Verwendung:
rule1 = Rule(
    rule_type="AND",
    left=Rule(
        rule_type="AND",
        left=Rule(
            rule_type="AND",
            left=Rule(raw_text="- Subjects who do"),
            right=Rule(
                rule_type="NOT",
                left=Rule(
                    rule_type="OR",
                    left=Rule(raw_text="not have malignant tissue available", condition=["malignant tissue"]),
                    right=Rule(raw_text="and accessible")
                )
            )
        ),
        right=Rule(
            rule_type="AND",
            left=Rule(raw_text="- The amount of excised malignant tissue is", condition=["malignant tissue"]),
            right=Rule(
                rule_type="NOT",
                left=Rule(
                    rule_type="OR",
                    left=Rule(raw_text="not sufficient for the ex vivo drug testing"),
                    right=Rule(raw_text="and/or genetic profiling.")
                )
            )
        )
    ),
    right=Rule(
        rule_type="OR",
        left=Rule(raw_text="- Patients with newly diagnosed tumors", condition=["tumors", "tumor"]),
        right=Rule(raw_text="and tumors that have high (>90%) cure rate with safe standard therapy.", condition=["tumors", "tumor"], observation=["cure rate"])
    )
)

# Ausgabe als Wörterbuch
import json
print(json.dumps(rule1.to_dict(), indent=4))
```
