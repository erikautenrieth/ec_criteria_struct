```python
import os
import shutil

def copy_files_and_verify(eval_root, n_root):
    for dirpath, dirnames, filenames in os.walk(eval_root):
        if filenames:  # Check if there are files in the current directory
            # Calculate the relative path from eval_root
            relative_path = os.path.relpath(dirpath, eval_root)
            # Determine the corresponding n_root path
            target_path = os.path.join(n_root, relative_path)

            # Ensure the target directory exists
            os.makedirs(target_path, exist_ok=True)

            # Copy all files to the target directory
            for file in filenames:
                src_file = os.path.join(dirpath, file)
                dest_file = os.path.join(target_path, file)
                shutil.copy2(src_file, dest_file)

            # Verify if the target directory has exactly 202 files
            if len(os.listdir(target_path)) == 202:
                print(f"Verified: {target_path} contains 202 files.")
            else:
                print(f"Warning: {target_path} does not contain 202 files.")


eval_root = "eval_n_shot"
n_root = "n_shot"
copy_files_and_verify(eval_root, n_root)
```
