```python
import pandas as pd

def read_evaluation_results(file_path):
    df = pd.read_excel(file_path, sheet_name='Model Results')
    return df

def find_low_tp_high_label(df, threshold=4):
    low_tp_high_label = {
        'AND': [],
        'OR': [],
        'NOT': []
    }

    for index, row in df.iterrows():
        if row['AND_label_count'] > threshold and row['AND_tp'] <= 1:
            low_tp_high_label['AND'].append(row['nct_number'])
        if row['OR_label_count'] > threshold and row['OR_tp'] <= 1:
            low_tp_high_label['OR'].append(row['nct_number'])
        if row['NOT_label_count'] > threshold and row['NOT_tp'] <= 1:
            low_tp_high_label['NOT'].append(row['nct_number'])

    return low_tp_high_label

# Beispielaufruf der Funktionen
file_path = 'excel_eval/Llama-3-70B-Instruct_5_shot_prompt_0_evaluation_results.xlsx'
df = read_evaluation_results(file_path)
low_tp_high_label_files = find_low_tp_high_label(df)

print("NCT files with high label count but low TP for AND:", low_tp_high_label_files['AND'])
print("NCT files with high label count but low TP for OR:", low_tp_high_label_files['OR'])
print("NCT files with high label count but low TP for NOT:", low_tp_high_label_files['NOT'])
```
