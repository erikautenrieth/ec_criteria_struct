```python
import os

def read_txt_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        content = f.read()
    return content

def get_word_offsets(text):
    words = text.split()
    offsets = []
    current_pos = 0

    for word in words:
        start_offset = text.find(word, current_pos)
        end_offset = start_offset + len(word)
        offsets.append((word, start_offset, end_offset))
        current_pos = end_offset

    return offsets

file_path = 'data/NCT03864393.txt'
text = read_txt_file(file_path)
word_offsets = get_word_offsets(text)

# Ausgabe der Wörter und ihrer Offsets
for word, start, end in word_offsets:
    print(f"Word: '{word}' (Offsets: {start}-{end})")
```
