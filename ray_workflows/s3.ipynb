```python
import boto3
from botocore.exceptions import NoCredentialsError

class S3FileManager:
    def __init__(self):
        self.s3_client = boto3.client('s3',
                                      endpoint_url='http://10.0.7.59:9000',
                                      aws_access_key_id='eautenrieth',
                                      aws_secret_access_key='Kennword1')

    def upload_to_s3(self, file_name, bucket, object_name):
        try:
            self.s3_client.upload_file(file_name, bucket, object_name)
            print(f"File {object_name} saved!")
            return True
        except NoCredentialsError:
            print("Credentials not available")
            return False
        except Exception as e:
            print(f"Error uploading file: {e}")
            return False

    def download_from_s3(self, bucket, object_name, file_name):
        try:
            self.s3_client.download_file(bucket, object_name, file_name)
            print(f"File {object_name} downloaded to {file_name}!")
            return True
        except NoCredentialsError:
            print("Credentials not available")
            return False
        except Exception as e:
            print(f"Error downloading file: {e}")
            return False
```

```python
s3Manager = S3FileManager()
```

```python
s3Manager.download_from_s3(bucket='raybucket', object_name="test.txt", file_name="./data/downloaded_test2.txt")
```
