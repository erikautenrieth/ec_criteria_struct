```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("darkgrid")


models_corrected = [
    "SciBERT (SOTA)",
    "Llama3-70B (2-Shot)",
    "Qwen2-72B (2-Shot)",
    "GPT-4o (2-Shot)",
    "Llama3-70B (10-Shot)",
    "Qwen2-72B (10-Shot)",
    "GPT-4o (10-Shot)",
    "Llama3-70B (Fine-Tuning)"
]

f1_scores_corrected = [75.9, 10.7, 12.8, 16.4, 14.7, 15.0, 18.3, 84.6]
def get_color(model):
    if "SOTA" in model:
        return '#1f77b4'  # Blau: typischer wissenschaftlicher Farbton
    elif "GPT-4o" in model:
        return '#d62728'  # Rot: symbolisch für "Closed"
    else:
        return '#2ca02c'  # Grün: für "Open"

colors_corrected = [get_color(model) for model in models_corrected]
models_corrected.reverse()
f1_scores_corrected.reverse()
colors_corrected.reverse()
fig, ax = plt.subplots(figsize=(12, 8))
bars = ax.barh(models_corrected, f1_scores_corrected, color=colors_corrected)
for bar in bars:
    ax.text(bar.get_width() + 1, bar.get_y() + bar.get_height() / 2, f'{bar.get_width():.1f}%', va='center', fontsize=12, fontweight='bold')

ax.set_xlabel('F1-Score (%)', fontsize=16, fontweight='bold')
ax.set_title('Evaluation der  Operatorenextraktion mit JSON-Ausgabe', fontsize=18, pad=40, fontweight='bold')
ax.set_xlim(0, 100)
ax.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()


ax.tick_params(axis='y', labelsize=16, width=2, labelcolor='k')
ax.tick_params(axis='x', labelsize=14)
legend_elements = [
    plt.Rectangle((0, 0), 1, 1, fc='#2ca02c', label='Open-Source'),
    plt.Rectangle((0, 0), 1, 1, fc='#d62728', label='Closed-Source'),
    plt.Rectangle((0, 0), 1, 1, fc='#1f77b4', label='SOTA')
]
ax.legend(handles=legend_elements, loc='center right', fontsize=12)
plt.savefig('op_eval.svg', format='svg', bbox_inches='tight')
plt.show()
```
