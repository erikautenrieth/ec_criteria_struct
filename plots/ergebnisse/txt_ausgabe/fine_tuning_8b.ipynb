```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Daten vorbereiten
data = {
    'r': [16, 32, 64, 128, 256, 512, 2048],
    'F1_r': [74.5, 74.5, 75.7, 77.1, 76.3, 76.1, 77.1],
    'alpha': [16, 32, 64, 128, 256, 512, 1024],
    'F1_alpha': [69.3, 71.4, 71.6, 75.0, 77.1, 74.4, 74.7],
    'epochs': [10, 20, 30, 40],
    'F1_epochs': [77.1, 74.6, 77.0, 75.6],
    'lr': [2e-4, 5e-6, 5e-5],
    'F1_lr': [50.7, 69.1, 75.5]
}


plt.figure(figsize=(15, 10))

# r optimierung
plt.subplot(2, 2, 1)
sns.lineplot(x=data['r'], y=data['F1_r'], marker='o')
plt.xscale('log', base=2)
plt.title('Optimierung von r')
plt.xlabel('r')
plt.ylabel('F1-Score (%)')

# alpha optimierung
plt.subplot(2, 2, 2)
sns.lineplot(x=data['alpha'], y=data['F1_alpha'], marker='o')
plt.xscale('log', base=2)
plt.title('Optimierung von alpha')
plt.xlabel('alpha')
plt.ylabel('F1-Score (%)')

# Epochen variation
plt.subplot(2, 2, 3)
sns.lineplot(x=data['epochs'], y=data['F1_epochs'], marker='o')
plt.title('Variation der Epochenanzahl')
plt.xlabel('Epochen')
plt.ylabel('F1-Score (%)')

# Lernrate
plt.subplot(2, 2, 4)
sns.lineplot(x=data['lr'], y=data['F1_lr'], marker='o')
plt.xscale('log')
plt.title('Optimierung der Lernrate')
plt.xlabel('Learning Rate')
plt.ylabel('F1 Score (%)')

plt.tight_layout()
plt.suptitle('Fine-Tuning Evaluation von Llama3 8B', fontsize=16)
plt.subplots_adjust(top=0.93)
plt.savefig('pics/fine_tuning_llama3_8B.png')
plt.show() 
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# Daten vorbereiten
data = {
    'Parameter': ['r'] * 7 + ['α'] * 7,
    'Value': [16, 32, 64, 128, 256, 512, 2048] + [16, 32, 64, 128, 256, 512, 1024],
    'F1': [74.5, 74.5, 75.7, 77.1, 76.3, 76.1, 77.1] + [69.3, 71.4, 71.6, 75.0, 77.1, 74.4, 74.7]
}

df = pd.DataFrame(data)

# Seaborn-Stil setzen
sns.set_theme(style="darkgrid", context="paper", font="sans-serif", font_scale=1.2)
plt.rcParams.update({
    "font.sans-serif": ["Helvetica", "Arial"],
    "font.family": "sans-serif",
    "axes.labelweight": "bold",
    "axes.titleweight": "bold",
    "figure.titleweight": "bold"
})

# Plot erstellen
plt.figure(figsize=(12, 8))
ax = sns.lineplot(data=df, x='Value', y='F1', hue='Parameter', marker='o', markersize=8, linewidth=2)

# Plot-Styling
ax.set_xscale('log', base=2)
ax.set_xlabel('Parameterwert', fontsize=14, labelpad=10)
ax.set_ylabel('F1-Score', fontsize=14, labelpad=10)
ax.set_title('Optimierung von r und α für Llama3 8B Fine-Tuning', fontsize=16, pad=20)

# Legende anpassen
plt.legend(title='Parameter', title_fontsize='12', fontsize='10', loc='lower right')
plt.grid(True, linestyle='--', alpha=0.6)
plt.ylim(65, 80)

# X-Achse Ticks anpassen
plt.xticks([16, 32, 64, 128, 256, 512, 1024, 2048], ['16', '32', '64', '128', '256', '512', '1024', '2048'])

# Werte annotieren
for line in ax.lines:
    if line.get_label() in ['r', 'α']:
        y = line.get_ydata()
        x = line.get_xdata()
        for i, txt in enumerate(y):
            ax.annotate(f'{txt:.1f}', (x[i], y[i]), xytext=(0, 5), textcoords='offset points', ha='center', va='bottom', fontsize=8)

for spine in ax.spines.values():
    spine.set_visible(True)

sns.despine()
plt.tight_layout()

plt.savefig('pics/lora_para.png', bbox_inches='tight')

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Data from the table
data = {
    'Epochs': [10, 10, 10, 10, 20, 40, 40],
    'r': [128, 128, 256, 256, 256, 128, 256],
    'alpha': [256, 512, 256, 512, 256, 256, 256],
    'F1': [77.3, 77.3, 78.0, 75.7, 75.3, 76.8, 75.1]
}

# Create DataFrame
df = pd.DataFrame(data)

sns.set(style="darkgrid")
# Create the plot
plt.figure(figsize=(12, 8))

# Custom color palette with only two colors
colors = ['#1f77b4', '#ff7f0e']

# Create the main plot
sns.lineplot(data=df, x="Epochs", y="F1", hue="r", style="alpha", markers=True, dashes=False, palette=colors)

# Customize the plot
plt.title("Llama3 70B Fine-Tuning", fontsize=18, fontweight='bold')
plt.xlabel("Number of Epochs", fontsize=14)
plt.ylabel("F1-Score", fontsize=14)

# Add a grid for better readability
plt.grid(True, linestyle='--', alpha=0.7)

# Customize the legend
legend = plt.legend(title='Parameters', title_fontsize='13', fontsize='11', bbox_to_anchor=(1.05, 1), loc='upper left')

# Update legend labels to include both r and alpha
new_labels = ['r=128, α=256', 'r=128, α=512', 'r=256, α=256', 'r=256, α=512']
for t, l in zip(legend.get_texts(), new_labels):
    t.set_text(l)

# Add annotations for key points
for i, row in df.iterrows():
    plt.annotate(f'{row["F1"]:.1f}', (row["Epochs"], row["F1"]),
                 xytext=(5, 5), textcoords='offset points', fontsize=9)

# Adjust the layout and save the figure
plt.tight_layout()
plt.savefig('pics/70b_fine_tuning_simplified.png', bbox_inches='tight')
plt.show()
```
