```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns


sns.set_theme(style="darkgrid")
sns.set_context("talk", font_scale=1.5)
shots = np.array([1, 2, 3, 4, 5])
precision = np.array([45.77, 48.17, 48.76, 50.90, 50.82])
precision_err = np.array([2.28, 1.66, 1.62, 1.48, 1.65])
recall = np.array([71.19, 70.85, 69.68, 69.03, 67.66])
recall_err = np.array([1.94, 2.44, 2.67, 3.09, 3.60])
f1 = np.array([51.87, 53.85, 54.13, 55.74, 55.41])
f1_err = np.array([1.96, 1.35, 1.72, 1.61, 2.14])
colors = sns.color_palette("Set1")
plt.figure(figsize=(12, 8))

# Funktionen für schattierte Fehlerbereiche
def plot_with_error_bands(x, y, yerr, color, label):
    plt.plot(x, y, '-o', color=color, label=label, markersize=8, linewidth=2.5)
    plt.fill_between(x, y - yerr, y + yerr, color=color, alpha=0.3)

# Plotten mit schattiertem Fehlerbereich
plot_with_error_bands(shots, precision, precision_err, 'blue', 'Precision')
plot_with_error_bands(shots, recall, recall_err, 'orange', 'Recall')
plot_with_error_bands(shots, f1, f1_err, colors[2], 'F1-Score')
plt.xlabel('Anzahl der Shots', fontweight='bold', fontsize=16, labelpad=10)
plt.ylabel('Score (%)', fontweight='bold', fontsize=16, labelpad=10)
plt.title('Llama3-70B Performance mit k-Shot', fontweight='bold', fontsize=20, pad=20)
plt.legend(loc='lower right', frameon=True, fancybox=True, shadow=True, fontsize=14)
plt.xticks(shots, fontsize=16, fontweight='bold')
plt.ylim(40, 75)
plt.yticks(np.arange(40, 80, 5), fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
for spine in plt.gca().spines.values():
    spine.set_visible(True)
    spine.set_linewidth(2)
plt.tight_layout()
plt.savefig('pics/n_shot.svg', format='svg', bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns


sns.set_theme(style="darkgrid")
sns.set_context("talk", font_scale=1.5)
labels = ['5-Zufall', '5-Auswahl', '5-Max', '5-Min']
precision = np.array([50.82, 48.66, 49.88, 47.51])
precision_err = np.array([1.65, 0.59, 1.23, 0.67])
recall = np.array([67.66, 70.28, 75.27, 65.35])
recall_err = np.array([3.60, 2.46, 1.39, 1.28])
f1 = np.array([55.41, 54.34, 56.32, 51.91])
f1_err = np.array([2.14, 1.21, 1.10, 0.86])
colors = sns.color_palette("colorblind")

plt.figure(figsize=(12, 8))
x = range(len(labels))

def plot_with_error_bands(x, y, yerr, color, label):
    plt.plot(x, y, '-o', color=color, label=label, markersize=8, linewidth=2)
    plt.fill_between(x, y - yerr, y + yerr, color=color, alpha=0.3)


plot_with_error_bands(x, precision, precision_err, 'blue', 'Precision')
plot_with_error_bands(x, recall, recall_err, 'orange', 'Recall')
plot_with_error_bands(x, f1, f1_err, colors[2], 'F1-Score')
plt.xlabel('5-Shot Varianten', fontweight='bold', fontsize=16, labelpad=15)
plt.ylabel('Score (%)', fontweight='bold', fontsize=16, labelpad=15)
plt.title('Llama3-70B Performance  mit 5-Shot Varianten', fontweight='bold', fontsize=20, pad=20)
plt.legend(loc='upper right', frameon=True, fancybox=True, shadow=True, fontsize=14)
plt.xticks(x, labels, fontweight='bold', fontsize=14, rotation=45, ha='right')
plt.ylim(40, 80)
plt.yticks(np.arange(40, 85, 5), fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
for spine in plt.gca().spines.values():
    spine.set_visible(True)
    spine.set_linewidth(2)

plt.tight_layout()
plt.savefig('pics/5shot_variants.svg', format='svg', bbox_inches='tight')
plt.show()
```
