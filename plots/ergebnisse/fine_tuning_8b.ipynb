```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Daten vorbereiten
data = {
    'r': [16, 32, 64, 128, 256, 512, 2048],
    'F1_r': [74.5, 74.5, 75.7, 77.1, 76.3, 76.1, 77.1],
    'alpha': [16, 32, 64, 128, 256, 512, 1024],
    'F1_alpha': [69.3, 71.4, 71.6, 75.0, 77.1, 74.4, 74.7],
    'epochs': [10, 20, 30, 40],
    'F1_epochs': [77.1, 74.6, 77.0, 75.6],
    'lr': [2e-4, 5e-6, 5e-5],
    'F1_lr': [50.7, 69.1, 75.5]
}

# Plot erstellen
plt.figure(figsize=(15, 10))

# r optimierung
plt.subplot(2, 2, 1)
sns.lineplot(x='r', y='F1_r', data=data, marker='o')
plt.xscale('log', base=2)
plt.title('Optimierung von r')
plt.xlabel('r (log scale)')
plt.ylabel('F1 Score')

# alpha optimierung
plt.subplot(2, 2, 2)
sns.lineplot(x='alpha', y='F1_alpha', data=data, marker='o')
plt.xscale('log', base=2)
plt.title('Optimierung von alpha')
plt.xlabel('alpha (log scale)')
plt.ylabel('F1 Score')

# Epochen variation
plt.subplot(2, 2, 3)
sns.lineplot(x='epochs', y='F1_epochs', data=data, marker='o')
plt.title('Variation der Epochenanzahl')
plt.xlabel('Epochen')
plt.ylabel('F1 Score')

# Lernrate
plt.subplot(2, 2, 4)
sns.lineplot(x='lr', y='F1_lr', data=data, marker='o')
plt.xscale('log')
plt.title('Optimierung der Lernrate')
plt.xlabel('Lernrate (log scale)')
plt.ylabel('F1 Score')

plt.tight_layout()
plt.suptitle('Fine-Tuning Evaluation von Llama3 8B', fontsize=16)
plt.subplots_adjust(top=0.93)
plt.show()
```
