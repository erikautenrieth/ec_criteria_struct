```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Daten vorbereiten
data = {
    'Model': ['SciBERT', 'R-BERT + SciBERT', 'Naive greedy match', 'Llama2 70B (3-Shot)', 'Llama3 8B (Zero-Shot)', 'Llama3 70B (Zero-Shot)', 'Gemma 27B (5-Shot)', 'Llama3 8B (5-Shot)', 'Llama3 70B (5-Shot)', 'Qwen2 72B (5-Shot)', 'Qwen2 72B (15-Shot)', 'Llama3 8B (Fine-Tuned)', 'Llama3 70B (Fine-Tuned)', 'GPT-4o (Zero-Shot)', 'GPT-4o (5-Shot)', 'GPT-4o (15-Shot)', 'GPT-4o mini (5-Shot)', 'GPT-4o mini (15-Shot)', 'Claude 3.5 Sonnet (5-Shot)', 'Claude 3.5 Sonnet (15-Shot)'],
    'F1': [75.9, 75.0, 50.9, 15.3, 27.5, 49.7, 41.7, 33.5, 55.8, 49.6, 56.3, 77.1, 78.0, 46.8, 52.9, 55.2, 53.5, 55.6, 56.5, 60.7],
    'Type': ['SOTA', 'SOTA', 'Baseline', 'Open Source', 'Open Source', 'Open Source', 'Open Source', 'Open Source', 'Open Source', 'Open Source', 'Open Source', 'Open Source', 'Open Source', 'Closed Source', 'Closed Source', 'Closed Source', 'Closed Source', 'Closed Source', 'Closed Source', 'Closed Source']
}

df = pd.DataFrame(data)

# Plot erstellen
plt.figure(figsize=(20, 12))

# F1 Score Vergleich
ax = sns.barplot(data=df, x='Model', y='F1', hue='Type', dodge=False)
plt.title('Auswertung der Operatorenextraktion', fontsize=20, pad=20)
plt.xlabel('Modell', fontsize=18, labelpad=15, fontweight='bold')
plt.ylabel('F1-Score (%)', fontsize=18, labelpad=15, fontweight='bold')
plt.xticks(rotation=45, ha='right', fontsize=12, fontweight='bold')
plt.yticks(fontsize=12)
plt.legend(title='Modelltyp', title_fontsize='14', fontsize='12')

# Achsen fett machen
for spine in ax.spines.values():
    spine.set_linewidth(2)

# Hinzufügen von Wertbeschriftungen
for i, v in enumerate(df['F1']):
    plt.text(i, v + 0.5, f'{v:.1f}%', ha='center', va='bottom', fontsize=10)

plt.tight_layout()

# Als PDF speichern
plt.savefig('operator_auswertung.pdf', format='pdf', bbox_inches='tight')

plt.show()
```
