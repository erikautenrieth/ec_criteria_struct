```python
import json
import os

# Funktion zum Einlesen und Verarbeiten von JSON-Dateien aus einem Ordner
def process_json_files(input_directory, output_directory):
    # Sicherstellen, dass der Ausgabeordner existiert
    os.makedirs(output_directory, exist_ok=True)

    # Iteriere über alle Dateien im Eingabeverzeichnis
    for filename in os.listdir(input_directory):
        if filename.endswith(".json"):
            input_file_path = os.path.join(input_directory, filename)
            output_file_path = os.path.join(output_directory, filename)

            # JSON-Datei laden
            with open(input_file_path, "r", encoding="utf-8") as input_file:
                data = json.load(input_file)

            # Verarbeiten der Daten: Ersetzen von Typen und Entfernen von arg2_text bei NOT
            processed_data = []
            for item in data:
                if "type" in item and "arg1_text" in item:
                    new_item = {
                        "type": item["type"].replace("Or", "OR").replace("And", "AND").replace("Negation", "NOT"),
                        "arg1_text": item["arg1_text"]
                    }
                    if new_item["type"] != "NOT" and "arg2_text" in item:
                        new_item["arg2_text"] = item["arg2_text"]
                    processed_data.append(new_item)

            # Ergebnisse in der Ausgabedatei speichern
            with open(output_file_path, "w", encoding="utf-8") as output_file:
                json.dump(processed_data, output_file, indent=4)

            print(f"Verarbeitete Datei gespeichert: {output_file_path}")

# Hauptfunktion zum Ausführen des gesamten Prozesses
def main():
    input_directory = "label"
    output_directory = "output"
    process_json_files(input_directory, output_directory)

# Ausführen der Hauptfunktion
if __name__ == "__main__":
    main()
```
