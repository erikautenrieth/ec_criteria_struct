```python
import os
import json

p = 2

input_dir = f"p{p}/"
output_dir= f"p{p}_model_input/"
os.makedirs(output_dir, exist_ok=True)
for file_name in os.listdir(input_dir):
    print(file_name)
    if file_name.endswith(f"_parsed_{p}.json"):
        input_file_path = os.path.join(input_dir, file_name)
        with open(input_file_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
        json_string = json.dumps(data, ensure_ascii=True,indent=2, separators=(',', ':')) #separators=(',', ':')
        with open(output_dir+file_name.replace(".json", ".txt"), 'w') as file:
            file.write(json_string)
```
