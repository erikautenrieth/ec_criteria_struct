```python
import os
import json
```

```python
def parse_logic(logic):
    if 'raw_text' in logic:
        return logic['raw_text']
    for key in logic:
        if key in ('AND', 'OR', 'NOT'):
            operator = key
            operands = logic[key]
            if operator == 'NOT':
                return f"NOT({parse_logic(operands['left'])})"
            elif operator in ('AND', 'OR'):
                left = parse_logic(operands['left']) if 'left' in operands else ''
                right = parse_logic(operands['right']) if 'right' in operands else ''
                return f"{operator}({left}, {right})"

    return ""
```

```python
input_dir = "p2/"
output_dir= "p2_formatted/"
for file_name in os.listdir(input_dir):
    if file_name.endswith("_parsed_2.json"):
        input_file_path = os.path.join(input_dir, file_name)
        with open(input_file_path, 'r', encoding='utf-8') as f:
            data = json.load(f)

        parsed_logic = parse_logic(data)

        output_file_name = file_name.replace("_parsed_2.json", "_formatted.txt")
        output_file_path = os.path.join(output_dir, output_file_name)

        # Speichere den formatierten logischen Ausdruck in der Ausgabedatei
        with open(output_file_path, 'w', encoding='utf-8') as f:
            f.write(parsed_logic)

        print(f"Processed and saved: {output_file_path}")
```
