```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Pfad zu den CSV-Dateien
csv_files = {
    'Llama-3-70B-Instruct': 'evaluate/batch1/Llama-3-70B-Instruct_3_shot_metrics_summary.csv',
    'Llama-3-8B-Instruct': 'evaluate/batch1/Llama-3-8B-Instruct_3_shot_metrics_summary.csv',
}

# Funktion zum Einlesen der CSV-Dateien
def load_metrics(csv_file, model_name):
    df = pd.read_csv(csv_file)
    df['Model'] = model_name
    return df

# Metriken für alle Modelle einlesen und kombinieren
metrics_df_list = []
for model, file in csv_files.items():
    df = load_metrics(file, model)
    metrics_df_list.append(df)

metrics_df = pd.concat(metrics_df_list)

```

```python
metrics_df
```

```python
# Metriken für AND, OR, NOT, DEPTH
metrics_to_plot = ['accuracy', 'precision', 'recall', 'f1_score']
pct_metrics_to_plot = ['pct_greater', 'pct_less', 'pct_equal']

# Reshape DataFrame for plotting
metrics_long_df = metrics_df.melt(id_vars=['Model', 'Metric'], value_vars=metrics_to_plot, var_name='Metric_Type', value_name='Value')

# Visualisierung der Metriken
for metric in metrics_to_plot:
    plt.figure(figsize=(14, 8))
    sns.barplot(x='Metric', y='Value', hue='Model', data=metrics_long_df[metrics_long_df['Metric_Type'] == metric])
    plt.title(f'Model Comparison for {metric.capitalize()}')
    plt.ylabel(f'{metric.capitalize()} Score')
    plt.xlabel('Metric Type')
    plt.legend(title='Model')
    plt.show()

# Heatmaps der prozentualen Differenzen
for pct_metric in pct_metrics_to_plot:
    pivot_df = metrics_df.pivot_table(index='Model', columns='Metric', values=pct_metric)
    plt.figure(figsize=(14, 8))
    sns.heatmap(pivot_df, annot=True, cmap='coolwarm', center=50)
    plt.title(f'Percentage {pct_metric.capitalize()} Comparison by Model and Metric')
    plt.xlabel('Metric Type')
    plt.ylabel('Model')
    plt.show()
```
