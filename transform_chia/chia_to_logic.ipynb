```python
def find_word_offsets(text):
    offsets = []
    start = 0

    while start < len(text):
        if text[start].isspace():
            start += 1
            continue

        end = start
        while end < len(text) and not text[end].isspace():
            end += 1

        offsets.append((start, end))
        start = end

    return offsets

with open(ec_file, 'r', encoding='utf-8') as file:
    content = file.read()
print(ec_file)
word_offsets = find_word_offsets(content)
for start, end in word_offsets:
    print(f"Start: {start}, End: {end}, Word: '{content[start:end]}'")
```
