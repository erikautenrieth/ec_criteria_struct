```python
from functions import *
import json
from Node import *
```

```python
type = "IC"
```

```python
exc_files = extract_unique_ids(directory='input', type="exc")
```

```python
inc_files = extract_unique_ids(directory='input', type="inc")
```

```python
for file in inc_files:
    ann_file = f'input/{file}.ann'
    ec_file = f'input/{file}.txt'
    output_file = f'output/p1/{file}_parsed'
    parsed_1 = output_file + "_1.json"

    ann_data = parse_ann_file(ann_file)
    offsets = replace_ids_with_offsets(ann_data)
    offsets = remove_last_elements(offsets)
    content = insert_character_at_offsets(ec_file, offsets)
    data = criteria_to_json(content, output_file=parsed_1, type=type)
    print(data)

    new_ann_data = extract_main_entities(ann_data)
    data = read_json(parsed_1)

    tree = build_tree(data)
    tree_json = tree.to_dict()
    ausgabe = json.dumps(tree_json, indent=2)
    save_json_to_file(ausgabe, f'output/p2/{file}_parsed_2.json')
    
    ausgabe = read_json(f'output/p2/{file}_parsed_2.json')
    update_conditions(ausgabe, new_ann_data)
    out = json.dumps(ausgabe, indent=2)
    save_json_to_file(out, f'output/p3/{file}_parsed_3.json')
    
    print(f"end {file}")
```
