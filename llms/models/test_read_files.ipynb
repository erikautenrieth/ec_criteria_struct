```python
from llms.models.helper_functions import read_text_file
import os

def read_all_files_from_directory(directory_path):
    files_content = {}
    try:
        for file_name in os.listdir(directory_path):
            file_path = os.path.join(directory_path, file_name)
            content = read_text_file(file_path)
            key = os.path.splitext(file_name)[0]
            files_content[key] = content
    except FileNotFoundError:
        print(f"Das Verzeichnis {directory_path} wurde nicht gefunden.")
    except Exception as e:
        print(f"Ein Fehler ist aufgetreten: {e}")

    return files_content


directory = "../input/chia"
contents = read_all_files_from_directory(directory)
```
