```python
folder_path = "output"
subfolders = ["p1", "p1_model_input", "p2", "p2_model_input", "p3", "p3_model_input"]
for subfolder in subfolders:
    os.makedirs(os.path.join(folder_path, subfolder), exist_ok=True)
```

```python
exc_files = extract_unique_ids(directory='input/ann_text', type="exc")
inc_files = extract_unique_ids(directory='input/ann_text', type="inc")
```

```python
Type = "IC" # IC, EC
for file in inc_files: #exc_files  inc_files
    ann_file = f'input/ann_text/{file}.ann'
    ec_file = f'input/ann_text/{file}.txt'
    output_file = f'{folder_path}/p1_getestet/{file}_parsed'
    parsed_1 = output_file + "_1.json"
    # p1
    ann_data = parse_ann_file(ann_file)
    offsets = replace_ids_with_offsets(ann_data)
    offsets = remove_last_elements(offsets)
    content = insert_character_at_offsets(ec_file, offsets)
    data = read_json(parsed_1)
    # Save P1 model input
    with open(f'{folder_path}/p1_model_input/{file}.txt', 'w', encoding="utf-8") as f:
         f.write("\n".join(data.values()))
    # p2
    tree = build_tree(data)
    tree_json = tree.to_dict()
    ausgabe = json.dumps(tree_json, indent=2)
    save_json_to_file(ausgabe, f'{folder_path}/p2/{file}.json')
    # Model p2 input
    save_json_to_txt(ausgabe, f'{folder_path}/p2_model_input/{file}.txt')
    save_json_to_txt_same_format(ausgabe, f'{folder_path}/p2_model_input/{file}.txt')
    
    new_ann_data = extract_main_entities(ann_data)
    ausgabe = read_json(f'{folder_path}/p2/{file}.json')
    update_conditions(ausgabe, new_ann_data)
    out = json.dumps(ausgabe, indent=2)
    save_json_to_file(out, f'{folder_path}/p3/{file}.json')
    # Model p3 input
    save_json_to_txt(out, f'{folder_path}/p3_model_input/{file}_parsed_3.txt')
```
