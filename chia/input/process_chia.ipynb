```python
import os

# Pfade für den Eingabe- und Ausgabeordner
input_folder = 'chia_text_half'
output_folder = 'chia_text_full'

# Sicherstellen, dass der Ausgabeordner existiert
os.makedirs(output_folder, exist_ok=True)

# Dateien im Eingabeordner durchlaufen
for filename in os.listdir(input_folder):
    # Nur Dateien betrachten, die mit "NCT" beginnen
    if filename.startswith("NCT") and filename.endswith(("_inc.txt", "_exc.txt")):
        # NCT-Nummer extrahieren
        nct_number = filename.split('_')[0]
        inc_file = os.path.join(input_folder, f"{nct_number}_inc.txt")
        exc_file = os.path.join(input_folder, f"{nct_number}_exc.txt")

        # Output-Dateipfad
        output_file = os.path.join(output_folder, f"{nct_number}.txt")

        # Inhalte zusammenführen
        with open(output_file, 'w', encoding='utf-8') as outfile:
            # Inclusion Criteria
            outfile.write("Inclusion Criteria:\n")
            if os.path.exists(inc_file):
                with open(inc_file, 'r', encoding='utf-8') as inc:
                    outfile.write(inc.read())

            # Exclusion Criteria
            outfile.write("\nExclusion Criteria:\n")
            if os.path.exists(exc_file):
                with open(exc_file, 'r', encoding='utf-8') as exc:
                    outfile.write(exc.read())

print(f"Die Dateien wurden erfolgreich in den Ordner '{output_folder}' gemergt.")
```
