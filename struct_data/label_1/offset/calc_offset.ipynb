```python
def calculate_offsets(text):
    lines = text.split('\n')
    current_offset = 0
    results = []

    for line in lines:
        if line.strip():  # Ignoriere leere Zeilen
            start_offset = current_offset
            end_offset = start_offset + len(line) - 1  # -1, da das Ende des Strings zu dem Offset gehört
            results.append((line.strip(), start_offset, end_offset))
            current_offset += len(line) + 1  # +1 für das Zeilenendezeichen

    return results
```

```python
criteria_text = """
Inclusion Criteria:
Overweight or obese subjects [according to body mass index (BMI)]
Fasting plasma glucose value between 100 and 125 mg/dl, with impaired fasting glucose or impaired glucose tolerance confirmed with oral glucose tolerance test (OGTT)
Total cholesterol values ≥ 200 mg/dl

Exclusion Criteria:
Patients with neoplastic and liver diseases, renal failure
Patients with type 1 or 2 diabetes mellitus
Pregnant or breastfeeding women
Hypersensitivity to any of the ingredients
Therapy with lipid-lowering drugs
Use of products containing red yeast rice
"""
```

```python
offsets = calculate_offsets(criteria_text)

for description, start, end in offsets:
    print(f"{description} // Offset: [{start}-{end}]")
```
