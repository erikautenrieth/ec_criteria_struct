## Parse Chia dataset

```python
type = "EC"
```

```python
exc_files = extract_unique_ids(directory='input', type="exc") 
```

```python
inc_files = extract_unique_ids(directory='input', type="inc")
```

```python
for file in exc_files:
    ann_file = f'input/{file}.ann'
    ec_file = f'input/{file}.txt'
    output_file = f'output/{file}_parsed'
    
    ann_data = parse_ann_file(ann_file)
    offsets = replace_ids_with_offsets(ann_data)
    offsets = remove_last_elements(offsets)
    content = insert_character_at_offsets(ec_file, offsets)
    data = criteria_to_json(content, output_file=output_file+"_1.json", type=type)
    parsed_data = parse_to_structured_json(data,type=type)
    save_json_to_file(parsed_data, output_file+"_2.json")
    print(f"end {file}")
```
