```python
import json
import gzip
def compress_json(input_data, output_file):
    # Entfernen von unnötigen Leerzeichen und Zeilenumbrüchen
    compact_data = json.dumps(input_data, separators=(',', ':'))

    # Speichern der kompakten Daten in einer gzip-Datei
    with gzip.open(output_file, 'wt', encoding='utf-8') as f:
        f.write(compact_data)

# Beispielaufruf
compress_json(ausgabe, 'NCT00050349_logic.json.gz')
```
