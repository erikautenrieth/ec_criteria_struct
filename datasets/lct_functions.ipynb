```python
import os
import shutil

# Pfad zum Verzeichnis, das durchsucht werden soll
source_directory = 'LCT'
destination_directory = 'LCT/ann_files'

# Sicherstellen, dass das Zielverzeichnis existiert
os.makedirs(destination_directory, exist_ok=True)

# Durch alle Dateien und Unterordner im Quellverzeichnis iterieren
for root, dirs, files in os.walk(source_directory):
    for file in files:
        if file.endswith('.ann'):
            # Vollständigen Pfad der Quelldatei erstellen
            source_file = os.path.join(root, file)
            # Vollständigen Pfad der Zieldatei erstellen
            destination_file = os.path.join(destination_directory, file)
            # Datei kopieren
            shutil.copy2(source_file, destination_file)
            print(f'Kopiert: {source_file} -> {destination_file}')

print('Alle .ann Dateien wurden erfolgreich kopiert.')
```
